```python
import random

def getrandomnumber():
    return random.randrange(1,10)

def accuracy(number, guess):
    if guess > number + 3 or guess < number - 3:
        return "Cold"
    elif guess == number:
        return "Right"
    else:
        return "Hot"
def runguess():
    sceretNumber = getrandomnumber()
    while True:
        userGuess = int(input("Enter the Guessed Number!"))

        hint = accuracy(sceretNumber, userGuess)
        if hint == "Right":
            print("Congralutaions! Your Guess is Correct")
            break
        else:
            print(hint)

if __name__ == '__main__':
    runguess()
```
